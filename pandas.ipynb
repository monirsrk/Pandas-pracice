```python
def extra(txt):
  data=''
  check=False
  for i in txt:
      if i=='(':
        check=True
        continue # Start capturing after '('
      if i==')':
        check=False
        break # Stop capturing after ')'
      if check==True: # Only append if inside parentheses
          data=data+i
  return data
```
